### 广播机制

#### 一维数组广播

```python
[0,1,2,3] * 3
```

```python
import numpy as np
# sort 排序
# [0,1,2,3]*3
arr1 = np.sort(np.array([0,1,2,3]*3)).reshape(4,3) #shape(4,3)
display(arr1)
arr2 = np.array([1,2,3]) # shape(3,)
display(arr2)
arr3 = arr1 + arr2 # arr2进行广播复制4份 shape(4,3)
arr3 # jupyter会将变量输出【Pycharm没有这样操作】
```

```python
display(arr2)
```

#### 二维数组广播

```python
import numpy as np
arr1 = np.sort(np.array([0,1,2,3]*3)).reshape(4,3) # shape(4,3)
# arr2，二维
arr2 = np.array([[1],[2],[3],[4]]) # shape(4,1)
display(arr1,arr2)
arr3 = arr1 + arr2 # arr2 进行广播复制3份 shape(4,3)
display(arr1,arr2,arr3)
```

#### 高维数组广播

```python
import numpy as np
arr1 = np.array([0,1,2,3,4,5,6,7]*3).reshape(3,4,2) #shape(3,4,2)
# arr2 = np.array([0,1,2,3,4,5,6,7]).reshape(4,2) #shape(4,2)
# arr2 = np.random.randint(0,10,size = (3,4,1))
arr2 = np.random.randint(0,10,size = (3,1,2))
display(arr1.shape,arr2.shape)
arr3 = arr1 + arr2 # arr2数组在0维上复制3份 shape(3,4,2)
arr3
```

#### where函数使用


```python
import numpy as np
arr1 = np.array([1,3,5,7,9])
arr2 = np.array([2,4,6,8,10])
cond = np.array([True,False,True,True,False])
np.where(cond,arr1,arr2) # True选择arr1，False选择arr2的值
# 输出 array([ 1,  4,  5,  7, 10])
```

```python
arr3 = np.random.randint(0,30,size = 20)
display(arr3)
np.where(arr3 < 15,arr3,-15) # 小于15还是自身的值，大于15设置成-15
```

#### 集合运算

```python
A = np.array([2,4,6,8])
B = np.array([3,4,5,6])
print(np.intersect1d(A,B)) # 交集 array([4, 6])
print(np.union1d(A,B)) # 并集 array([2, 3, 4, 5, 6, 8])
print(np.setdiff1d(A,B)) #差集，A中有，B中没有 array([2, 8])
```

### 线性代数

```python
#矩阵的乘积
A = np.array([[4,2,3],
              [1,3,1]]) # shape(2,3)
B = np.array([[2,7],
              [-5,-7],
              [9,3]]) # shape(3,2)
np.dot(A,B) # 矩阵运算 A的最后一维和B的第一维必须一致
```

```python
A @ B # 符号 @ 表示矩阵乘积运算
```

```python
#计算矩阵的逆
from numpy.linalg import inv,det,eig,qr,svd
A = np.array([[1,2,3],
              [2,3,4],
              [4,5,8]]) # shape(3,3)
B = inv(A) # 逆矩阵
display(B)
display(A @ B)
det(A)#计算矩阵行列式
```
