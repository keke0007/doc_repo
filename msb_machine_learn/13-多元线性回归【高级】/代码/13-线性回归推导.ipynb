### 正规方程

#### 二元一次方程

$W = (X^TX)^{-1}X^Ty$

```python
# 线性回归，到底是什么呢？
# 可以解决什么问题呢？
import numpy as np
X = np.array([[1,1],[2,-1]])
y = np.array([14,10])

# 正规方程使用 linalg 线性代数
w = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)
w
```

```python
from sklearn.linear_model import LinearRegression # 线性回归

model = LinearRegression(fit_intercept=True) # 默认这个参数是True，一般情况下， 使用True
model.fit(X,y) # 建模，训练

# coef斜率，线性方程，系数 == 斜率，截距
# 上面的方程（求解方程的变量，系数，斜率 + 截距），有没有截距？
display(model.coef_, model.intercept_)
```

```python
# # 带截距
# # x + y = 14
# -0.8 + 1.6 + 13.2 = 14

# # 2x - y = 10
# 2*-0.8 - 1.6 + 13.2 = 10
```

#### 八元一次方程

```python
# 上面八元一次方程对应的X数据
X = np.array([[  0 ,14 , 8 ,  0 ,  5,  -2,   9,  -3],
 [ -4 , 10 ,  6 ,  4 ,-14 , -2 ,-14  , 8],
 [ -1 , -6  , 5 ,-12 ,  3 , -3 ,  2 , -2],
 [  5 , -2  , 3 , 10  , 5 , 11 ,  4  ,-8],
 [-15 ,-15  ,-8 ,-15 ,  7 , -4, -12 ,  2],
 [ 11 ,-10 , -2 ,  4  , 3 , -9 , -6 ,  7],
 [-14 ,  0 ,  4 , -3  , 5 , 10 , 13 ,  7],
 [ -3 , -7 , -2 , -8  , 0 , -6 , -5 , -9]])
# 对应的y
y = np.array([ 339 ,-114  , 30 , 126, -395 , -87 , 422, -309])
display(X,y)
```

```python
# x1 到 x8分别是：1.,  5., 15.,  3.,  8.,  4., 17., 12.
w = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)
w
```

```python
model = LinearRegression(fit_intercept=False)
model.fit(X,y)
display(model.coef_,model.intercept_)
```

```python
y_new = y + 64
y_new
```

```python
# 思考一个问题，增加了截距，相当于b ，未知数
w = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y_new)
w
```

```python
# 有唯一解，需要增加一行
# 构建方程，解
w = np.array([ 1.,  5., 15.,  3.,  8.,  4., 17., 12.])
b = 64
X9 = np.random.randint(-10,10,size = 8) # x1 ~ x8【方程任意】
y9 = w.dot(X9) + b # 得到第 9个方程的目标值

# 合并数据
y_new = np.concatenate([y_new,[y9]])
y_new # 9个目标值了
```

```python
# 执行一次
X_new = np.concatenate([X_new,[np.concatenate([X9,[1]])]])
X_new
```

```python
model = LinearRegression(fit_intercept=False)

model.fit(X_new,y_new)
display(model.coef_,model.intercept_)
```

```python
model = LinearRegression(fit_intercept=True)

model.fit(X_new,y_new)
display(model.coef_,model.intercept_)
```
