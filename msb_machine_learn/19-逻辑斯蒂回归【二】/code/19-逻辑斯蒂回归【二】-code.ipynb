### 概率的计算

#### 二分类

```python
import numpy as np
from sklearn import datasets
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

X,y = datasets.load_iris(return_X_y=True)

cond = y != 2 # 将0和1筛选
X = X[cond]
y = y[cond]

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.15)
display(X.shape,X_train.shape,X_test.shape,y)
```

```python
model = LogisticRegression()
model.fit(X_train,y_train)
```

```python
y_ = model.predict(X_test)
print('算法预测：',y_)
print('真实结果：',y_test)
```

```python
proba_ = model.predict_proba(X_test) # 算法计算的概率
proba_
```

```python
proba_.argmax(axis = 1) # 找到最大值的索引
```

概率手动计算

```python
w_ = model.coef_
b_ = model.intercept_
display(w_,b_)
```

和训练数据，进行多元线性回归的计算

```python
X_test.shape
```

```python
w_.shape
```

```python
# 手动计算的！！！
z = X_test.dot(w_.T) + b_

def sigmoid(z):
    return 1/(1 + np.exp(-z))

# 自己算出来的概率，只有一列，默认情况下，计算的类别为1的概率
# 正样本，阳性。医院，清一色阴性，化验，
# negative ,阴性，0
# positive ，阳性，1 往往很多问题，都是关注阳性，所以，代码中返回的也是1，阳性【默认】
# 代码中，对1，特别关注
p = sigmoid(z) # 概率，就计算出来

np.concatenate([1- p,p],axis = 1) # 类别0 的概率
```

#### 多分类-ovr

```python
import numpy as np
from sklearn import datasets
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

X,y = datasets.load_iris(return_X_y=True)

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.15)
display(X.shape,X_train.shape,X_test.shape,y)
```

```python
np.set_printoptions(suppress=True)
```

```python
proba_ = model.predict_proba(X_test)
proba_
```

```python
proba_.argmax(axis = 1) # 根据概率，进行类别划分
```

计算3分类的概率

```python
w_ = model.coef_
b_ = model.intercept_
display(w_,b_)
```

```python
# 手动计算的！！！
z = X_test.dot(w_.T) + b_

def sigmoid(z):
    return 1/(1 + np.exp(-z))
# 自己算出来的概率，只有一列，默认情况下，计算的类别为1的概率
# 正样本，阳性。医院，清一色阴性，化验，
# negative ,阴性，0
# positive ，阳性，1 往往很多问题，都是关注阳性，所以，代码中返回的也是1，阳性【默认】
# 代码中，对1，特别关注
p = sigmoid(z) # 概率，就计算出来
p/(p.sum(axis = 1)).reshape(-1,1)
```

#### 多分类-multinomial

```python
import numpy as np
from sklearn import datasets
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

X,y = datasets.load_iris(return_X_y=True)

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.15)
display(X.shape,X_train.shape,X_test.shape,y)
```

```python
model = LogisticRegression(multi_class='multinomial')
model.fit(X_train,y_train)
```

```python
y_ = model.predict(X_test)
display(y_,y_test)
```

```python
model.predict_proba(X_test)
```

手动计算概率，softmax函数

```python
w_ = model.coef_
b_ = model.intercept_
display(w_,b_)
```

```python
np.set_printoptions(suppress=True)
z = X_test.dot(w_.T) + b_

def softmax(z):
    return np.exp(z)/(np.exp(z).sum(axis = 1)).reshape(-1,1)

softmax(z)
```
