### 线性回归-预测房价

```python
from sklearn.linear_model import LinearRegression

from sklearn import datasets

from sklearn.model_selection import train_test_split
```

```python
import warnings
warnings.filterwarnings('ignore')
```

加载波士顿房价【数据类型无所谓：癌症、糖尿病、身高】

```python
data = datasets.load_boston()
data
```

提取数据

```python
X = data['data'] # 特征，影响房价的指标

y = data['target'] # 房价
display(X.shape,y.shape,X[:5],y[:5])
```

拆分数据

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 50)
display(X_train.shape,X_test.shape)
```

建模和预测

```python
model = LinearRegression(fit_intercept=True)# 正规方程，封装好了

model.fit(X_train,y_train)# 训练，建模
```

```python
y_ = model.predict(X_test) # 保留的50个样本
display('算法预测的房价值：',y_.round(1))

display('真实房价是多少：',y_test)
```

模型评估

```python
model.score(X_test,y_test)
```

```python
# 不是死记硬背，根据数据提取 规律，方程，并不是所有的数据，都吻合
model.score(X_train,y_train)
```

```python
from sklearn.metrics import mean_squared_error # 均方误差
```

```python
# 均方误差，越小越好！
mean_squared_error(y_test,y_)
```

```python
# 模型非常好，预测非常准，完全一样！
# y_test == y_ 根据这个公式进行计算，结果是：0
((y_test - y_)**2).mean()
```

模型上线，模型部署

### 代码模拟梯度下降

```python
import numpy as np
import matplotlib.pyplot as plt
f = lambda x : (x - 3.5)**2 - 4.5*x + 10 # 一元二次
# 导函数 ，梯度
def g(x):
    return 2*(x - 3.5) - 4.5
eta = 0.1
# 最优解，x = 5.75
theta = 20 # 随机猜
for i in range(50):
    theta = theta - eta * g(theta)
    print(theta)
```

```python
import numpy as np
import matplotlib.pyplot as plt
f = lambda x : (x - 3.5)**2 - 4.5*x + 10 # 一元二次
# 导函数 ，梯度
def g(x):
    return 2*(x - 3.5) - 4.5
eta = 0.1 # 学习率，道士下山，步幅
# 最优解，x = 5.75
theta = -5 # 随机猜

last_theta = theta + 0.1 #记录一下，上一次权重多少 
while True:
    if np.abs(last_theta - theta) < 0.001:# 精确度，满足一定误差
        break
    last_theta = theta # 当前的theta赋值给了上一步的last_theta
    theta = theta - eta * g(theta) # 更新,又走了一步【新的】
    print(theta)
```
