### 加载数据

```python
import numpy as np
import pandas as pd

job = pd.read_csv('./job.csv')

display(job.shape,job.head())
```

### 重复数据删除

```python
job.head(15)
```

```python
cond = job.duplicated()
job[cond]
```

```python
# 删除的是重复行
job.drop_duplicates(inplace = True) # 重复数据， 某一行和其他一行，完全相同
job.shape
```

### 薪资数据转变【范围--->平均值】

```python
# 提取薪资，赋值修改原数据
job['salary'] = job['salary'].str.lower().str.extract(r'(\d+)k-(\d+)k').astype(np.int16).mean(axis =1)
job.head()
```

### 过滤非数据分析岗位

```python
cond = job['positionName'].str.contains('数据分析')
# 数据筛选
job = job[cond]
```

### job_detail中提取技能

```python
# 工作详细描述中，大写变成小写
job['job_detail'] = job['job_detail'].str.lower()
```

```python
job['Python'] = job['job_detail'].map(lambda x:1 if ('python' in x) else 0)
job["SQL"] = job["job_detail"].map(lambda x:1 if ('sql' in x) or ('hive' in x)  else 0)
job["Tableau"] = job["job_detail"].map(lambda x:1 if 'tableau' in x  else 0)
job["Excel"] = job["job_detail"].map(lambda x:1 if 'excel' in x  else 0)
job['SPSS/SAS'] = job['job_detail'].map(lambda x:1 if ('spss' in x) or ('sas' in x) else 0)
job.head()
```

### 行业信息处理

```python
job.shape
```

```python
def convert(item):
    items = item.split(',')
    if items[0] == '移动互联网' and len(items) > 1:
        return items[1]
    else:
        return items[0]
job['industryField'] = job['industryField'].map(convert)
# 保存
job.to_csv('./job_process.csv',index =False)
```

### 技能要求

```python
def get_level(x):
    if x["Python"] == 1:
        x["skill"] = "Python"
    elif x["SQL"] == 1:
        x["skill"] = "SQL"
    elif x["Excel"] == 1:
        x["skill"] = "Excel"
    elif x['SPSS/SAS'] == 1:
        x['skill'] = 'SPSS/SAS'
    else:
        x["skill"] = "其他"
    return x
job = job.apply(get_level,axis=1) # 数据转换

# 获取主要技能
x = job.loc[job.skill!='其他'][['salary','skill']]
cond1 = x['skill'] == 'Python'
cond2 = x['skill'] =='SQL'
cond3 = x['skill'] == 'Excel'
cond4 = x['skill'] == 'SPSS/SAS'

plt.figure(figsize=(12,8))
plt.title(label='             不同技能的薪资水平对比            ', 
          fontsize=32, weight='bold', color='white', 
          backgroundcolor='#c5b783',pad = 30)

plt.boxplot(x = [job.loc[job.skill!='其他']['salary'][cond1],
                job.loc[job.skill!='其他']['salary'][cond2],
                job.loc[job.skill!='其他']['salary'][cond3],
                job.loc[job.skill!='其他']['salary'][cond4]],
           vert = False,labels = ["Python/R","SQL","Excel",'SPSS/SAS'])

plt.tick_params(axis="both",labelsize=16)
plt.grid(axis = 'x',linewidth = 0.75)
plt.xticks(np.arange(0,61,10), [str(i)+"k" for i in range(0,61,10)])
plt.box(False)
plt.xlabel('工资', fontsize=18)
plt.ylabel('技能', fontsize=18)
plt.savefig('./技能要求.png')
```

### 不同规模公司招聘人才工作年限差别

```python
from matplotlib import gridspec
workYear_map = {
    "5-10年": 5,
    "3-5年": 4,
    "1-3年": 3,
    "1年以下": 2,
    "应届毕业生": 1}
color_map = {
    5:"#ff0000",
    4:"#ffa500",
    3:"#c5b783",
    2:"#3c7f99",
    1:"#0000cd"}

cond = job.workYear.isin(workYear_map)
job = job[cond] # 取出特定的工作年限的数据
job['workYear'] = job.workYear.map(workYear_map)
# 根据companySize进行排序，人数从多到少
job['companySize'] = job['companySize'].astype('category') # 数据类型，调整一下：类别
list_custom = ['2000人以上', '500-2000人','150-500人','50-150人','15-50人','少于15人']
job['companySize'] = job['companySize'].cat.reorder_categories(list_custom)
job.sort_values(by = 'companySize',inplace = True,ascending = False)

plt.figure(figsize=(12,11))
gs = gridspec.GridSpec(10,1)# 10行1列
plt.subplot(gs[:8]) # 取前8行，左闭右开，8取不到
plt.suptitle(t='            不同规模公司的用人需求差异          ', 
         fontsize=32, 
         weight='bold', color='white', backgroundcolor='#3c7f99')

plt.scatter(job.salary,job.companySize,
            c = job.workYear.map(color_map),
            s = (job.workYear*100),alpha = 0.35) # 外圈
plt.scatter(job.salary,job.companySize,
            c = job.workYear.map(color_map)) # 内圈
plt.grid(axis = 'x')
plt.xticks(np.arange(0,161,10), [str(i)+"k" for i in range(0,161,10)])
plt.xlabel('工资', fontsize=18)
plt.box(False)
plt.tick_params(labelsize = 18)


# 绘制底部标记
plt.subplot(gs[8:]) # 切片，gs,获取最后一个子视图，最后一行
x = np.arange(5)[::-1]
y = np.zeros(len(x))
s = x*100
plt.scatter(x,y,s=s,c=color_map.values(),alpha=0.3) # 外圈
plt.scatter(x,y,c=color_map.values()) # 内圈
plt.box(False)
plt.xticks(ticks=x,labels=list(workYear_map.keys()),fontsize=14)
plt.yticks(np.arange(1),labels=['  经验：'],fontsize=18)
plt.savefig('./不同规模公司招聘薪资工作经验差异.png')
```
